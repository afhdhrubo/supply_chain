## 3.4 Feature Hygiene: Automatic ID / Date / High-Cardinality Detection  *(new)*
A 53-column, order-level e-commerce dataset like this one has several columns that are neither leakage
nor useful predictors: order/customer/product identifiers, and free-text/very-high-cardinality fields
that would either blow up one-hot encoding or contribute no generalizable signal. This cell detects and
drops them **automatically by pattern and cardinality**, as a safety net beyond the explicit list in
Section 3, and prints exactly what it removed and why (for your Data Validation subsection).

```python
# ---------------- FEATURE HYGIENE CONFIGURATION ----------------
ID_NAME_PATTERN = re.compile(r'(^|[ _])(id)([ _]|$)', re.IGNORECASE)
PII_NAME_PATTERN = re.compile(r'email|password|street|image|description', re.IGNORECASE)
HIGH_CARDINALITY_THRESHOLD = 50   # object columns with more unique values than this are dropped
ID_UNIQUENESS_RATIO = 0.5         # object/int columns with unique-value ratio above this are ID-like

hygiene_dropped = {"id_like": [], "pii_like": [], "datetime_like": [], "high_cardinality": []}

feature_cols_pre = [c for c in df.columns if c != TARGET_COL]

for c in list(feature_cols_pre):
    if c not in df.columns:
        continue
    name_is_id = bool(ID_NAME_PATTERN.search(c))
    name_is_pii = bool(PII_NAME_PATTERN.search(c))
    uniqueness_ratio = df[c].nunique(dropna=True) / max(len(df), 1)

    if name_is_pii:
        hygiene_dropped["pii_like"].append(c)
        df = df.drop(columns=[c])
        continue
    if name_is_id and uniqueness_ratio > ID_UNIQUENESS_RATIO:
        hygiene_dropped["id_like"].append(c)
        df = df.drop(columns=[c])
        continue

print("Dropped as PII/free-text:", hygiene_dropped["pii_like"])
print("Dropped as ID-like (name pattern + high uniqueness):", hygiene_dropped["id_like"])
```

### 3.4-extra Constant and exact-duplicate columns  *(new)*
Two more automatic, dataset-agnostic checks: a column with **zero variance** (every row identical)
carries no information and should be dropped; a column that is an **exact duplicate** of another column
(e.g. two differently named fields that always hold the same value) inflates multicollinearity and
double-counts the same signal without adding anything. Both are detected and dropped generically here,
which caught real cases in the DataCo file (an all-constant `Product Status` field, and two duplicate
column pairs) without needing to hardcode their names.

```python
remaining_feature_cols = [c for c in df.columns if c != TARGET_COL]

constant_cols = [c for c in remaining_feature_cols if df[c].nunique(dropna=False) <= 1]
df = df.drop(columns=constant_cols)
hygiene_dropped["constant"] = constant_cols
print("Dropped constant (zero-variance) columns:", constant_cols)

remaining_feature_cols = [c for c in df.columns if c != TARGET_COL]
seen_hashes = {}
duplicate_cols = []
for c in remaining_feature_cols:
    col_hash = pd.util.hash_pandas_object(df[c], index=False).sum()
    if col_hash in seen_hashes and df[c].equals(df[seen_hashes[col_hash]]):
        duplicate_cols.append(c)
    else:
        seen_hashes[col_hash] = c

df = df.drop(columns=duplicate_cols)
hygiene_dropped["exact_duplicate"] = duplicate_cols
print("Dropped exact-duplicate columns (identical to another retained column):", duplicate_cols)
```

### 3.4a Datetime detection
Any remaining object column that parses as a date for the large majority of its (non-null) values is
treated as a raw datetime string. **Order-date-like columns are engineered into safe features in 3.4b
before being dropped here; anything shipping-date-like is dropped outright** (only knowable after the
order has actually shipped, i.e. after the outcome).

```python
remaining_object_cols = df.select_dtypes(include=["object"]).columns.tolist()
remaining_object_cols = [c for c in remaining_object_cols if c != TARGET_COL]

datetime_like_cols = []
for c in remaining_object_cols:
    sample = df[c].dropna().sample(min(500, df[c].notna().sum()), random_state=RANDOM_STATE) if df[c].notna().sum() > 0 else df[c]
    if len(sample) == 0:
        continue
    parsed = pd.to_datetime(sample, errors="coerce", format="mixed")
    if parsed.notna().mean() > 0.9:
        datetime_like_cols.append(c)

print("Detected datetime-like columns:", datetime_like_cols)

ORDER_DATE_COL = next((c for c in datetime_like_cols if re.search(r'order.*date', c, re.IGNORECASE)), None)
SHIP_DATE_COLS = [c for c in datetime_like_cols if re.search(r'ship', c, re.IGNORECASE) and c != ORDER_DATE_COL]

print("Order-date column (will be feature-engineered, then dropped raw):", ORDER_DATE_COL)
print("Shipping-date-like column(s) (dropped outright -- post-outcome leakage risk):", SHIP_DATE_COLS)
```

### 3.4b Safe, order-time-only date feature engineering
Extracts day-of-week, month, quarter, and weekend indicators from the **order date only** -- these are
genuinely known at the moment the order is placed, before any disruption/late-delivery outcome exists, so
they carry legitimate signal (e.g. seasonal or weekday demand/capacity effects) without leaking the
outcome.

```python
if ORDER_DATE_COL is not None:
    parsed_order_date = pd.to_datetime(df[ORDER_DATE_COL], errors="coerce", format="mixed")
    df["order_day_of_week"] = parsed_order_date.dt.dayofweek
    df["order_month"] = parsed_order_date.dt.month
    df["order_quarter"] = parsed_order_date.dt.quarter
    df["order_is_weekend"] = parsed_order_date.dt.dayofweek.isin([5, 6]).astype(int)
    print("Engineered 4 order-time-only date features:",
          ["order_day_of_week", "order_month", "order_quarter", "order_is_weekend"])
else:
    print("No order-date column detected -- skipping date feature engineering.")

cols_to_drop_dates = [c for c in datetime_like_cols if c in df.columns]
df = df.drop(columns=cols_to_drop_dates)
hygiene_dropped["datetime_like"] = cols_to_drop_dates
print("Dropped raw datetime string columns:", cols_to_drop_dates)
```

### 3.4c High-cardinality categorical auto-drop
One-hot encoding a categorical column with hundreds/thousands of unique values (e.g. product name, city)
would explode the feature space and add noise rather than signal at this sample size. These are dropped
and reported; a natural extension for future work is target/frequency encoding instead of dropping them
outright.

```python
remaining_object_cols = [c for c in df.select_dtypes(include=["object"]).columns if c != TARGET_COL]
high_card_cols = [c for c in remaining_object_cols if df[c].nunique(dropna=True) > HIGH_CARDINALITY_THRESHOLD]

df = df.drop(columns=high_card_cols)
hygiene_dropped["high_cardinality"] = high_card_cols
print(f"Dropped {len(high_card_cols)} high-cardinality categorical columns (> {HIGH_CARDINALITY_THRESHOLD} unique values):")
print(high_card_cols)

with open(f"{METRIC_DIR}/feature_hygiene_report.json", "w") as f:
    json.dump(hygiene_dropped, f, indent=2)

FEATURE_COLS = [c for c in df.columns if c != TARGET_COL]
NUMERIC_COLS = df[FEATURE_COLS].select_dtypes(include=[np.number]).columns.tolist()
CATEGORICAL_COLS = [c for c in FEATURE_COLS if c not in NUMERIC_COLS]

print(f"\nAfter hygiene: {len(FEATURE_COLS)} features retained "
      f"({len(NUMERIC_COLS)} numeric, {len(CATEGORICAL_COLS)} categorical).")
print("Numeric:", NUMERIC_COLS)
print("Categorical:", CATEGORICAL_COLS)
```

## 3.5 Leakage & Multicollinearity Audit
Same automatic, data-driven audit used on the previous dataset: univariate ROC-AUC (numeric) / Cramer's V
(categorical) against the target, thresholded and auto-dropped, plus a VIF multicollinearity report. This
is what would catch anything the explicit list in Section 3 missed (e.g. a slightly differently named
column that still leaks the label).

```python
LEAKAGE_AUC_THRESHOLD  = 0.97
LEAKAGE_CORR_THRESHOLD = 0.90
LEAKAGE_WHITELIST = []

def _cramers_v(confusion_matrix):
    chi2 = stats.chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum()
    r, k = confusion_matrix.shape
    denom = min(k - 1, r - 1)
    if denom <= 0 or n == 0:
        return 0.0
    return float(np.sqrt((chi2 / n) / denom))

leakage_scores = {}

for c in NUMERIC_COLS:
    col = df[c]
    if col.nunique(dropna=True) < 2:
        continue
    try:
        filled = col.fillna(col.median())
        auc = roc_auc_score(df[TARGET_COL], filled)
        leakage_scores[c] = max(auc, 1 - auc)
    except Exception:
        pass

for c in CATEGORICAL_COLS:
    try:
        ct = pd.crosstab(df[c].fillna("missing"), df[TARGET_COL])
        leakage_scores[c] = _cramers_v(ct.values)
    except Exception:
        pass

leakage_series = pd.Series(leakage_scores).sort_values(ascending=False)
print("Univariate association with target (AUC for numeric / Cramer's V for categorical):")
print(leakage_series.round(4))

LEAK_SUSPECT_COLS = [c for c in NUMERIC_COLS
                      if leakage_scores.get(c, 0) >= LEAKAGE_AUC_THRESHOLD and c not in LEAKAGE_WHITELIST]
LEAK_SUSPECT_COLS += [c for c in CATEGORICAL_COLS
                       if leakage_scores.get(c, 0) >= LEAKAGE_CORR_THRESHOLD and c not in LEAKAGE_WHITELIST]

print("\nColumns flagged as target leakage:", LEAK_SUSPECT_COLS if LEAK_SUSPECT_COLS else "none")

leakage_report = {
    "auc_threshold": LEAKAGE_AUC_THRESHOLD,
    "cramers_v_threshold": LEAKAGE_CORR_THRESHOLD,
    "scores": {k: float(v) for k, v in leakage_scores.items()},
    "dropped_columns": LEAK_SUSPECT_COLS,
}
with open(f"{METRIC_DIR}/leakage_audit_report.json", "w") as f:
    json.dump(leakage_report, f, indent=2)
```

```python
# Multicollinearity audit (VIF) -- report only, no auto-drop (trees are robust to it)
from statsmodels.stats.outliers_influence import variance_inflation_factor

numeric_pre_leakage_removal = [c for c in NUMERIC_COLS if c not in LEAK_SUSPECT_COLS]
vif_rows = []
if len(numeric_pre_leakage_removal) > 1:
    X_vif = df[numeric_pre_leakage_removal].fillna(df[numeric_pre_leakage_removal].median())
    for i, col in enumerate(numeric_pre_leakage_removal):
        try:
            vif_val = variance_inflation_factor(X_vif.values, i)
        except Exception:
            vif_val = np.nan
        vif_rows.append({"feature": col, "VIF": vif_val})

vif_df = pd.DataFrame(vif_rows).sort_values("VIF", ascending=False)
vif_df.to_csv(f"{TABLE_DIR}/table00b_multicollinearity_vif.csv", index=False)
print("Variance Inflation Factors (VIF > 10 suggests notable multicollinearity):")
print(vif_df.round(2))
HIGH_VIF_COLS = vif_df.loc[vif_df["VIF"] > 10, "feature"].tolist()
```

## 3.6 Leakage Ablation  *(new -- quantifies the audit's impact, for your manuscript)*
Trains a quick, single Random Forest **with** the flagged leakage columns still included vs. **without**
them (everything else identical), on an 80/20 split. Reporting this side by side turns "we removed some
columns because we suspected leakage" into a quantified, citable result: exactly how much apparent
performance the leakage was buying, and what genuinely predictive performance looks like once it's gone.
This table is exactly the kind of evidence a Q1 reviewer wants to see in a Data Validation subsection.

```python
def quick_eval(feature_cols, numeric_cols, categorical_cols, label):
    Xa = df[feature_cols]
    ya = df[TARGET_COL]
    Xa_tr, Xa_te, ya_tr, ya_te = train_test_split(Xa, ya, test_size=0.2, stratify=ya, random_state=RANDOM_STATE)

    prep = ColumnTransformer(transformers=[
        ("num", Pipeline([("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())]), numeric_cols),
        ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")), ("onehot", OneHotEncoder(handle_unknown="ignore"))]), categorical_cols),
    ])
    pipe = Pipeline([("prep", prep), ("clf", RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1))])
    pipe.fit(Xa_tr, ya_tr)
    proba = pipe.predict_proba(Xa_te)[:, 1]
    pred = (proba >= 0.5).astype(int)
    return {
        "label": label, "n_features": len(feature_cols),
        "Accuracy": accuracy_score(ya_te, pred), "F1": f1_score(ya_te, pred),
        "ROC_AUC": roc_auc_score(ya_te, proba), "PR_AUC": average_precision_score(ya_te, proba),
    }

results_with_leak = quick_eval(FEATURE_COLS, NUMERIC_COLS, CATEGORICAL_COLS, "WITH flagged leakage columns")

clean_features = [c for c in FEATURE_COLS if c not in LEAK_SUSPECT_COLS]
clean_numeric = [c for c in NUMERIC_COLS if c not in LEAK_SUSPECT_COLS]
clean_categorical = [c for c in CATEGORICAL_COLS if c not in LEAK_SUSPECT_COLS]
results_without_leak = quick_eval(clean_features, clean_numeric, clean_categorical, "WITHOUT flagged leakage columns")

ablation_df = pd.DataFrame([results_with_leak, results_without_leak]).set_index("label").round(4)
ablation_df.to_csv(f"{TABLE_DIR}/table00c_leakage_ablation.csv")
print(ablation_df)

if not LEAK_SUSPECT_COLS:
    print("\nNo columns were flagged in Section 3.5, so both rows above should be identical -- "
          "this is itself a useful sanity check that the audit is behaving correctly.")

# Apply the drop for the rest of the notebook
df = df.drop(columns=LEAK_SUSPECT_COLS)
FEATURE_COLS, NUMERIC_COLS, CATEGORICAL_COLS = clean_features, clean_numeric, clean_categorical
print(f"\nProceeding with {len(FEATURE_COLS)} leakage-audited features "
      f"({len(NUMERIC_COLS)} numeric, {len(CATEGORICAL_COLS)} categorical).")
```

### SHAP speed fix (added after the Section 10 cell stalled for 2+ hours)
If `BEST_MODEL_NAME` is `RandomForest` trained with `max_depth=None`, its trees can be extremely deep and
leafy on a ~100k-row training set, and SHAP's exact `TreeExplainer` algorithm has complexity roughly
`O(trees x leaves x depth^2)` **per explained row** -- for XGBoost/LightGBM/CatBoost (all capped at
`max_depth=6` in this notebook) this finishes in well under a minute even on thousands of rows, but for an
unbounded-depth Random Forest with 500 trees it can genuinely take hours. Two changes address this without
requiring you to retrain anything:
1. `SHAP_SAMPLE_SIZE` is reduced here to a much smaller value -- time scales roughly linearly with sample
   size, so far fewer rows is directly proportional to time saved. A few hundred rows is still enough for a
   stable global SHAP summary plot.
2. Section 10's `explainer.shap_values(...)` call now passes `approximate=True`, which uses a much faster
   approximate (Saabas-style) computation instead of the exact Shapley algorithm -- this is the change that
   actually removes the depth-squared blowup.

If SHAP is still too slow after this, the most robust fix is to retrain Random Forest with a depth cap
(e.g. `max_depth=25`) -- ask if you'd like an updated Section 7 that does this.

`CalibratedClassifierCV` fitting already happened in the previous run and is **not repeated** here --

and metrics from it.

```python
def expected_calibration_error(y_true, y_prob, n_bins=10):
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    bin_ids = np.clip(np.digitize(y_prob, bins) - 1, 0, n_bins - 1)
    ece = 0.0
    for b in range(n_bins):
        mask = bin_ids == b
        if mask.sum() == 0:
            continue
        acc = y_true[mask].mean()
        conf = y_prob[mask].mean()
        ece += (mask.sum() / len(y_true)) * abs(acc - conf)
    return ece

ece_raw = expected_calibration_error(y_test.values, proba_test_raw)
print(f"ECE before calibration: {ece_raw:.4f}")
```

### 9b. Bootstrap Uncertainty

**Speed fix (added because Section 9b was also taking a very long time):** the original
version refits the entire pipeline `N_BOOTSTRAP` times, resampling the **full** training set (~108k rows)
on every replicate. For XGBoost/LightGBM/CatBoost that is slow but usually tolerable; for an
unbounded-depth Random Forest -- the same root cause as the Section 10 SHAP stall -- each individual
refit can itself take minutes, so 30-150 replicates compounds into hours. Three changes below fix this
without changing what the bootstrap estimate measures: (1) **subsample** a fraction of the training set
per replicate instead of a full resample (a standard, valid bootstrap variant known as "subagging"),
(2) **fewer replicates**, still enough for a stable standard-deviation estimate, and (3) if the best model
is Random Forest with unbounded depth, its bootstrap clones only are capped at `max_depth=20` (the saved
best/calibrated model used everywhere else in the notebook is untouched). Progress now also prints
elapsed time and an ETA every replicate, instead of every 10, so it's clear it's actually moving.

```python
BOOTSTRAP_SUBSAMPLE_FRAC = 0.3   # fraction of the training set resampled per bootstrap replicate
N_BOOTSTRAP = 10 if QUICK_MODE else 20   # reduced from 30 / 150

print(f"Bootstrap settings: N_BOOTSTRAP={N_BOOTSTRAP}, "
      f"subsample fraction={BOOTSTRAP_SUBSAMPLE_FRAC} "
      f"(~{int(len(X_train) * BOOTSTRAP_SUBSAMPLE_FRAC):,} rows per replicate instead of {len(X_train):,}).")
```

```python
rng = np.random.RandomState(RANDOM_STATE)
bootstrap_probas = np.zeros((N_BOOTSTRAP, len(X_test)))
bootstrap_sample_size = max(1000, int(len(X_train) * BOOTSTRAP_SUBSAMPLE_FRAC))

# Bootstrap clones re-use the best model's hyperparameters, EXCEPT that an unbounded-depth Random
# Forest is capped at max_depth=20 for these refits only -- this only affects the bootstrap
# uncertainty estimate, not the saved best/calibrated model used everywhere else in the notebook.
bootstrap_clone_params = dict(models[BEST_MODEL_NAME].get_params())
if BEST_MODEL_NAME == "RandomForest" and bootstrap_clone_params.get("max_depth") is None:
    bootstrap_clone_params["max_depth"] = 20
    print("NOTE: capping RandomForest max_depth=20 for bootstrap refits only, to keep runtime bounded.")

bootstrap_start = time.time()
for b in range(N_BOOTSTRAP):
    idx = rng.choice(len(X_train), size=bootstrap_sample_size, replace=True)
    X_bs, y_bs = X_train.iloc[idx], y_train.iloc[idx]
    pipe_bs = Pipeline(steps=[
        ("prep", preprocessor),
        ("clf", models[BEST_MODEL_NAME].__class__(**bootstrap_clone_params)),
    ])
    pipe_bs.fit(X_bs, y_bs)
    bootstrap_probas[b, :] = pipe_bs.predict_proba(X_test)[:, 1]
    elapsed = time.time() - bootstrap_start
    avg_per_rep = elapsed / (b + 1)
    eta = avg_per_rep * (N_BOOTSTRAP - (b + 1))
    print(f"  bootstrap {b + 1}/{N_BOOTSTRAP} done | elapsed {elapsed/60:.1f} min | ETA {eta/60:.1f} min")

bootstrap_mean = bootstrap_probas.mean(axis=0)
bootstrap_std = bootstrap_probas.std(axis=0)
print(f"Mean bootstrap predictive std across test set: {bootstrap_std.mean():.4f}")
np.save(f"{METRIC_DIR}/bootstrap_probas.npy", bootstrap_probas)
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(bootstrap_std, bins=30, color="#55A868")
ax.set_xlabel("Bootstrap std of predicted probability (per shipment)")
ax.set_ylabel("Count")
ax.set_title("Distribution of Bootstrap Predictive Uncertainty")
plt.tight_layout()
plt.savefig(f"{FIGURE_DIR}/fig05_bootstrap_uncertainty_distribution.png", dpi=300)
plt.show()
```

empirical coverage **per subgroup** so the paper can explicitly claim (or caveat) uniform reliability

## 11. Decision Component: Risk-Aware Intervention Prioritization

```python
HIGH_THRESH = 0.70
LOW_THRESH = 0.30

def assign_tier(prob, is_ambiguous, high_thresh=HIGH_THRESH, low_thresh=LOW_THRESH):
    if is_ambiguous:
        return "Uncertain Risk - Human Review"
    if prob >= high_thresh:
        return "High Risk - Immediate Intervention"
    if prob < low_thresh:
        return "Low Risk - No Intervention"
    return "Uncertain Risk - Human Review"

decision_df = pd.DataFrame({
    "true_label": y_test.values,
    "calibrated_probability": proba_test_cal,
    "bootstrap_std": bootstrap_std,
    "conformal_ambiguous": conformal_ambiguous,
})
if SUBGROUP_COL is not None and SUBGROUP_COL in X_test.columns:
    decision_df[SUBGROUP_COL] = X_test[SUBGROUP_COL].values

# Real per-order economics carried through for Section 12 (business layer)
for econ_col in ["Sales", "Order Profit Per Order", "Order Item Profit Ratio"]:
    if econ_col in X_test.columns:
        decision_df[econ_col] = X_test[econ_col].values

decision_df["tier"] = [
    assign_tier(p, amb) for p, amb in zip(decision_df["calibrated_probability"], decision_df["conformal_ambiguous"])
]

tier_summary = decision_df["tier"].value_counts()
print(tier_summary)

fig, ax = plt.subplots(figsize=(7, 5))
tier_summary.plot(kind="bar", color=["#C44E52", "#DD8452", "#55A868"], ax=ax)
ax.set_ylabel("Number of shipments")
ax.set_title("Intervention Tier Assignment on Test Set")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.savefig(f"{FIGURE_DIR}/fig09_intervention_tiers.png", dpi=300)
plt.show()

decision_df.to_csv(f"{TABLE_DIR}/table04_decision_layer_output.csv", index=False)
```

## 12. Cost-Benefit & Business Decision Layer  *(now using real per-order economics)*
Unlike the earlier version of this notebook (which used flat, illustrative `COST_*` constants for every
shipment), this dataset provides **real order-level financial fields** (`Sales`, `Order Profit Per Order`).
The cost of an undetected disruption and the cost of a false alarm are now computed **per order, from that
order's actual economics**, rather than a single number applied uniformly. The one remaining assumption
-- what fraction of an order's value is actually put at risk by a late delivery (e.g., via a refund,
partial discount, or customer churn) -- is isolated into a single, clearly labeled, swept parameter
(`LATE_DELIVERY_RISK_FACTOR`) so it is fully transparent and testable in the sensitivity analysis below.
`COST_INTERVENTION` and `COST_HUMAN_REVIEW` remain illustrative operational placeholders (this dataset has
no field for actual expediting/staffing costs) -- replace them with your organization's real figures.

```python
ECON_COL = "Order Profit Per Order" if "Order Profit Per Order" in decision_df.columns else (
    "Sales" if "Sales" in decision_df.columns else None
)
print("Using economic field for order-level cost/benefit:", ECON_COL)

# ---------------- BUSINESS ASSUMPTIONS (edit to your case study) ----------------
COST_INTERVENTION = 25          # $ operational cost of proactively intervening on ONE order (illustrative)
COST_HUMAN_REVIEW = 5           # $ cost of a human analyst reviewing ONE "Uncertain" order (illustrative)
LATE_DELIVERY_RISK_FACTOR = 0.30  # fraction of an order's profit/value put at risk by a late delivery

def business_outcome(row, intervention_cost=None, human_review_cost=None, risk_factor=None):
    ic = COST_INTERVENTION if intervention_cost is None else intervention_cost
    hc = COST_HUMAN_REVIEW if human_review_cost is None else human_review_cost
    rf = LATE_DELIVERY_RISK_FACTOR if risk_factor is None else risk_factor

    order_value_at_risk = abs(row[ECON_COL]) * rf if ECON_COL is not None else 100.0  # fallback if no econ field
    tier, y_true = row["tier"], row["true_label"]

    if tier == "High Risk - Immediate Intervention":
        return ic if y_true == 1 else ic  # intervention is spent regardless; no separate false-alarm penalty beyond ic
    if tier == "Uncertain Risk - Human Review":
        return hc + (ic if y_true == 1 else 0)
    return order_value_at_risk if y_true == 1 else 0  # Low Risk, no intervention: full exposure if it actually disrupts

decision_df["policy_cost"] = decision_df.apply(business_outcome, axis=1)

cost_do_nothing = decision_df.apply(
    lambda r: (abs(r[ECON_COL]) * LATE_DELIVERY_RISK_FACTOR if ECON_COL else 100.0) if r["true_label"] == 1 else 0,
    axis=1,
).sum()
cost_intervene_all = len(decision_df) * COST_INTERVENTION
cost_policy = decision_df["policy_cost"].sum()

business_report = {
    "economic_field_used": ECON_COL,
    "late_delivery_risk_factor": LATE_DELIVERY_RISK_FACTOR,
    "n_test_orders": int(len(decision_df)),
    "actual_late_deliveries": int(decision_df["true_label"].sum()),
    "cost_do_nothing_baseline": float(cost_do_nothing),
    "cost_intervene_all_baseline": float(cost_intervene_all),
    "cost_under_proposed_policy": float(cost_policy),
    "savings_vs_do_nothing": float(cost_do_nothing - cost_policy),
    "savings_vs_intervene_all": float(cost_intervene_all - cost_policy),
    "roi_vs_do_nothing_pct": float(100 * (cost_do_nothing - cost_policy) / max(cost_do_nothing, 1)),
    "avg_risk_exposure_per_order": float(cost_policy / len(decision_df)),
}
print(json.dumps(business_report, indent=2))
with open(f"{METRIC_DIR}/business_costbenefit_report.json", "w") as f:
    json.dump(business_report, f, indent=2)
```

```python
fig, ax = plt.subplots(figsize=(10, 7))
bars = ["Do Nothing", "Intervene on All", "Proposed Policy"]
vals = [cost_do_nothing, cost_intervene_all, cost_policy]
ax.bar(bars, vals, color=["#C44E52", "#DD8452", "#55A868"])
ax.set_ylabel("Total cost on test set ($)")
ax.set_title("Business Cost Comparison Across Decision Strategies")
for i, v in enumerate(vals):
    ax.text(i, v, f"${v:,.0f}", ha="center", va="bottom")
plt.tight_layout()
plt.savefig(f"{FIGURE_DIR}/fig10_business_cost_comparison.png", dpi=300)
plt.show()
```

### 12a. Cost Sensitivity ("Tornado") Analysis

```python
SENSITIVITY_PCT_RANGE = [-0.30, -0.15, 0.0, 0.15, 0.30]
BASE_PARAMS = dict(intervention_cost=COST_INTERVENTION, human_review_cost=COST_HUMAN_REVIEW, risk_factor=LATE_DELIVERY_RISK_FACTOR)

def total_policy_cost(params):
    return decision_df.apply(lambda row: business_outcome(row, **params), axis=1).sum()

base_total = total_policy_cost(BASE_PARAMS)

sensitivity_rows = []
for param in BASE_PARAMS:
    for pct in SENSITIVITY_PCT_RANGE:
        varied = dict(BASE_PARAMS)
        varied[param] = BASE_PARAMS[param] * (1 + pct)
        total = total_policy_cost(varied)
        sensitivity_rows.append({"parameter": param, "pct_change": pct, "varied_value": varied[param],
                                  "total_policy_cost": total, "delta_vs_base": total - base_total})

sensitivity_df = pd.DataFrame(sensitivity_rows)
sensitivity_df.to_csv(f"{TABLE_DIR}/table06_cost_sensitivity.csv", index=False)
print(f"Base total policy cost: ${base_total:,.0f}")
sensitivity_df
```

```python
tornado = sensitivity_df.groupby("parameter")["total_policy_cost"].agg(["min", "max"])
tornado["range"] = tornado["max"] - tornado["min"]
tornado = tornado.sort_values("range")

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(tornado.index, tornado["max"] - tornado["min"], left=tornado["min"], color="#4C72B0")
ax.axvline(base_total, color="black", linestyle="--", linewidth=1, label=f"Base cost (${base_total:,.0f})")
ax.set_xlabel("Total policy cost on test set ($)")
ax.set_title("Sensitivity of Policy Cost to Business Assumptions (+/-30%)")
ax.legend()
plt.tight_layout()
plt.savefig(f"{FIGURE_DIR}/fig11_cost_sensitivity_tornado.png", dpi=300)
plt.show()
print("\nMost impactful assumption on total cost:", tornado["range"].idxmax())
```

### 12b. Cost-Optimal Decision Threshold

```python
threshold_grid = np.round(np.arange(0.50, 0.96, 0.05), 2)
threshold_costs = []
for th in threshold_grid:
    tiers_at_th = [assign_tier(p, amb, high_thresh=th) for p, amb in
                   zip(decision_df["calibrated_probability"], decision_df["conformal_ambiguous"])]
    tmp = decision_df.copy()
    tmp["tier"] = tiers_at_th
    total_cost_at_th = tmp.apply(business_outcome, axis=1).sum()
    threshold_costs.append({"HIGH_THRESH": th, "total_policy_cost": total_cost_at_th})

threshold_df = pd.DataFrame(threshold_costs)
threshold_df.to_csv(f"{TABLE_DIR}/table07_threshold_cost_sweep.csv", index=False)

optimal_row = threshold_df.loc[threshold_df["total_policy_cost"].idxmin()]
default_cost = threshold_df.loc[threshold_df["HIGH_THRESH"] == HIGH_THRESH, "total_policy_cost"]
default_cost = default_cost.iloc[0] if len(default_cost) else np.nan
print(f"Cost-minimizing HIGH_THRESH: {optimal_row['HIGH_THRESH']} "
      f"(total cost ${optimal_row['total_policy_cost']:,.0f} vs. ${default_cost:,.0f} at default HIGH_THRESH={HIGH_THRESH})")

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(threshold_df["HIGH_THRESH"], threshold_df["total_policy_cost"], marker="o", color="#55A868")
ax.axvline(optimal_row["HIGH_THRESH"], color="black", linestyle="--", linewidth=1,
           label=f"Cost-optimal threshold = {optimal_row['HIGH_THRESH']}")
ax.set_xlabel("HIGH_THRESH (intervention probability threshold)")
ax.set_ylabel("Total policy cost on test set ($)")
ax.set_title("Total Business Cost vs. Intervention Threshold")
ax.legend()
plt.tight_layout()
plt.savefig(f"{FIGURE_DIR}/fig12_threshold_cost_sweep.png", dpi=300)
plt.show()
```

## 14. Limitations (for the manuscript's Discussion section)
- **Feature hygiene and leakage thresholds are heuristic**, not a formal proof of zero leakage. Review the
  printed `leakage_series` (3.5) and `hygiene_dropped` report (3.4) manually against domain knowledge of
  what's genuinely known at order time before submission.
- **`LATE_DELIVERY_RISK_FACTOR` (Section 12) is still an assumption**, even though it's now applied to real
  per-order Sales/Profit figures rather than a flat constant -- it represents "what fraction of order value
  is actually at risk from a late delivery," which should ideally be calibrated against real churn/refund
  data if available, or clearly flagged as an illustrative multiplier if not.
- Conformal prediction's marginal coverage guarantee holds on average; Section 9c's subgroup coverage table
  should be discussed explicitly, especially if any market region under-covers by a meaningful margin.
- `N_BOOTSTRAP`/`N_CV_SPLITS` should be run with `QUICK_MODE = False` for final manuscript numbers.
- SHAP is computed on a random sample (`SHAP_SAMPLE_SIZE`) of the test set for tractability on this
  180k-row dataset -- standard practice, but should be stated explicitly in the methods.
- High-cardinality categorical columns (Section 3.4c) were dropped rather than target/frequency-encoded;
  future work could recover their signal with a leakage-safe encoding scheme (fit only on training folds).
- SHAP explanations describe correlational, model-internal feature attributions, not causal drivers of
  late delivery.
